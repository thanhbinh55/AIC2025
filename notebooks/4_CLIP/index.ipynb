```python
import os
import faiss
import numpy as np
from tqdm import tqdm
import glob
```

```python
feature_path = '../../dict/CLIPv2_features/L21_a/V001.npy'
features = np.load(feature_path)

print("Shape:", features.shape[1])
```

```python
features_dir = '../../dict/CLIPv2_features'

index = faiss.IndexFlatIP(features.shape[1])

for data_part in tqdm(sorted(os.listdir(features_dir))):
    for feature_path in tqdm(sorted(glob.glob(os.path.join(features_dir, data_part) +'/*.npy'))):
        feats = np.load(feature_path)
        for feat in feats:
            feat = feat.astype(np.float32).reshape(1,-1)
            index.add(feat)

faiss.write_index(index, f"../../dict/CLIPv2_cosine.bin")
print("Index written to CLIPv2_cosine.bin")
```
